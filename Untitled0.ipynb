```python
# This function simulates daily decisions of the GenZ character
def simulate_day():
    

def greet_user():
    print("Hey bestie! 💅 Welcome to the GenZ Life Simulator 💫", flush=True)
    name = input("What's your name? 👀: ")
    age = int(input("How old are you? 🎂: "))
    user = {"name": name, "age": age}
    print(f"\nNice to meet you, {name}! Let's vibe together ✨\n", flush=True)
    return user


def explore_library():
    print("\n📚 Welcome to the Explore Library! Let's find a book for you 😍", flush=True)
    topic = input("Choose a topic (science, fiction, history, motivation, ai): ").strip().lower()

    books = {
        "science": ["A Brief History of Time – Stephen Hawking", "The Selfish Gene – Richard Dawkins"],
        "fiction": ["1984 – George Orwell", "The Great Gatsby – F. Scott Fitzgerald"],
        "history": ["Sapiens – Yuval Noah Harari", "Guns, Germs, and Steel – Jared Diamond"],
        "motivation": ["Atomic Habits – James Clear", "The Power of Now – Eckhart Tolle"],
        "ai": ["Artificial Intelligence – Stuart Russell", "Deep Learning – Ian Goodfellow"]
    }

    if topic in books:
        print(f"\n✨ Based on your choice '{topic}', here are some book suggestions:", flush=True)
        for b in books[topic]:
            print("📖", b, flush=True)
    else:
        print("Oops, I don’t have books on that topic yet 😅", flush=True)


def update_stats(activity, stats):
    act = activity.lower()
    if act == "study":
        stats["energy"] -= 20
        stats["points"] += 10
        stats["mood"] = "Focused 🧠"
    elif act == "eat":
        stats["energy"] += 15
        stats["mood"] = "Satisfied 😋"
    elif act == "play":
        stats["energy"] -= 10
        stats["points"] += 5
        stats["mood"] = "Excited 🎮"
    elif act == "sleep":
        stats["energy"] = 100
        stats["mood"] = "Refreshed 😴"
    elif act == "explore library":
        explore_library()
    else:
        print("Invalid choice bestie 😭", flush=True)
    stats["energy"] = max(0, min(100, stats["energy"]))
    return stats


def show_status(stats):
    print("\n📊 Your Current Stats:", flush=True)
    for k, v in stats.items():
        print(f"{k.capitalize()}: {v}", flush=True)
    print("-------------------------", flush=True)


def end_game(stats):
    print("\n🎯 Game Over! Here's your summary:", flush=True)
    for k, v in stats.items():
        print(f"{k.capitalize()}: {v}", flush=True)
    print("Thanks for playing, bestie! 💅✨", flush=True)


def chat_mode():
    responses = {
        "hi": "Hey bestie! 😎",
        "how are you": "I’m vibing, wbu? 💅",
        "bye": "Aww, see you soon 😢"
    }
    print("\n💬 Let's Chat! Type 'bye' to go back to the menu.", flush=True)
    while True:
        user_input = input("You: ").lower().strip()
        if user_input == "bye":
            print("Returning to the main menu 💫", flush=True)
            break
        print("AI Bestie:", responses.get(user_input, "Hmm… idk what to say 🤔"), flush=True)


def play_game(user):
    stats = {"energy": 100, "points": 0, "mood": "Chill 😎"}
    print(f"\nWelcome {user['name']}! Let's start your GenZ Life Simulation 💫", flush=True)

    while True:
        print("\nChoose an activity:", flush=True)
        print("1. Study 📚", flush=True)
        print("2. Eat 🍔", flush=True)
        print("3. Play 🎮", flush=True)
        print("4. Sleep 😴", flush=True)
        print("5. Explore Library 📖", flush=True)
        print("6. Chat 💬", flush=True)
        print("7. Quit 🚪", flush=True)

        choice = input("Enter your choice: ").strip().lower()

        if choice in ("1", "study"):
            update_stats("study", stats)
        elif choice in ("2", "eat"):
            update_stats("eat", stats)
        elif choice in ("3", "play"):
            update_stats("play", stats)
        elif choice in ("4", "sleep"):
            update_stats("sleep", stats)
        elif choice in ("5", "explore library", "library", "explore"):
            update_stats("explore library", stats)
        elif choice in ("6", "chat"):
            chat_mode()
        elif choice in ("7", "quit"):
            end_game(stats)
            break
        else:
            print("Invalid input 😅 try again!", flush=True)

        show_status(stats)


def main():
    user = greet_user()
    play_game(user)


main()
```
